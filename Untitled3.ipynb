```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)
```

```python
(x_train , y_train),(x_test,y_test)=mnist.load_data()
print("training images shapes:",x_train.shape)
print("testing images shape:",x_test.shape)
```

```python
plt.figure(figsize=(10,4))
for i in range(10):
  plt.subplot(2,5,i+1)
  plt.imshow(x_train[i],cmap='gray')
  plt.title(f"lable:{y_train[i]}")
  plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
x_train_norm = x_train.astype('float32')/255.0
x_test_norm=x_test.astype('float32')/255.0
x_train_flat = x_train_norm.reshape(x_train_norm.shape[0],28*28)
x_test_flat = x_test_norm.reshape(x_test_norm.shape[0],28*28)
print("flattened training data shape:",x_train_flat.shape)
print("flattened testing data shape:",x_test_flat.shape)
```

```python
model=Sequential([
    Dense(32,activation='relu', input_shape=(784,)),
    Dense(10,activation='softmax')
])
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history=model.fit(
    x_train_flat, y_train,
    epochs=8,
    batch_size=128,
    validation_split=0.1
)
```
